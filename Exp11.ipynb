```python
import edge_tts

# Text to convert to speech
generated_text = """
Renewable energy is important because it helps reduce pollution,
decreases dependence on fossil fuels, and provides a cleaner and
more sustainable source of energy for the future.
"""

async def generate_audio(text, output_file):

    communicate = edge_tts.Communicate(
        text=text,
        voice="en-US-AriaNeural"
    )

    await communicate.save(output_file)


await generate_audio(
    generated_text,
    "content_audio.mp3"
)

print("Audio generated successfully!")
```
